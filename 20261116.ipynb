```python
수def odd_even(num):          #함수 정수
  if num % 2 == 0:
    return '짝수'
  else:
    return'홀수'

int_num = int(input('정수를 입력하세요: '))
print(odd_even(int_num))#함수 호출
```

```python
def tri_area(base, height):
  area = 0.5 * base * height
  return area


b = int(input('삼각형의 밑변을 입력하세요:'))
h = int(input('삼각형의 높이를 입력하세요:'))
result = tri_area(b,h)
print(f'삼각형의 넓이는 {result}입니다.')
```

```python
concat = lambda s1, s2: s1 + ' '+s2     #함다 함수
result = concat('파이썬', '프로그래밍')     #함수 호출
print('문자열 합치기:', result)
```
